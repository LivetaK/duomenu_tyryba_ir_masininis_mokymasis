```python
import pandas as pd
```

```python
df = pd.read_csv("norm_full.csv")
```

```python
def is_it_outlier(data):
    data = data.copy()
    columns = data.columns.tolist()
    data['is_outlier'] = False
    for col in columns:
        column = pd.to_numeric(data[col])
        Q1 = column.quantile(0.1)
        Q3 = column.quantile(0.9)
        IQR = Q3 - Q1
        lower_bound = Q1 - 3 * IQR
        upper_bound = Q3 + 3 * IQR

        mask = (column < lower_bound) | (column > upper_bound)
        data.loc[mask, 'is_outlier'] = True
    return data

temp_data = is_it_outlier(df)
mask = temp_data['is_outlier'] == True
print("viso duomenu:", temp_data.shape[0])
print("isskirciu skaicius:", mask.sum())
print("isskirciu procentas:", (mask.sum() / temp_data.shape[0]) * 100)
clean_all_data = temp_data.loc[~mask] # saliname isskirtis
```

```python
clean_all_data.drop(columns=['is_outlier'], inplace=True)

```
