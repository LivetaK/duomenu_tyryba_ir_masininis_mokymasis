```python
import pandas as pd
import numpy as np
data = pd.read_csv("aprasomoji/kmeans_clustered(10)_original_data.csv")
data.drop(columns=["label"], inplace=True)
```

```python
stats_list = []

clusters = data["cluster"].unique()
clusters.sort()

for cl in clusters:
    subset = data[data["cluster"] == cl]
    
    for col in data.columns:
        if col == "cluster":
            continue
        
        x = subset[col]
        stats = {
            "Klasteris": cl,
            "Variable": col,
            "Minimumas": np.nanmin(x),
            "Q1": np.nanquantile(x, 0.25),
            "Mediana": np.nanmedian(x),
            "Vidurkis": np.nanmean(x),
            "Q3": np.nanquantile(x, 0.75),
            "Maksimumas": np.nanmax(x),
            "Dispersija": np.nanvar(x, ddof=1),
        }
        stats_list.append(stats)

stats_df = pd.DataFrame(stats_list)
stats_df = stats_df.sort_values(["Variable", "Klasteris"]).reset_index(drop=True)

stats_df = stats_df.round(2)
```
